### Publication Age Analysis

**Hypothesis:** Books without assigned genres tend to be newer, explaining their lower rating count.

**Findings:**
- Books with genres: Median publication year is 2002.
- Books without genres: Median publication year is 1998.
- Books without genres are typically older, contradicting the original hypothesis.

**Conclusion:** The hypothesis was not supported. Publication age alone does not explain the engagement gap between books with and without genres.

For example, Dostoevsky's "White Nights" may have multiple editions, each with a unique book_id but the same work_id. This could explain why the dataset contains 38,321 repeated work_id occurrences despite having no duplicate book_id values.

```python
df_full[
    df_full["work_id"].notna() &
    df_full["work_id"].duplicated(keep=False)
][["title", "first_author", "work_id", "book_id", "format"]].sort_values("work_id").head(15) 
# inspect duplicate work_id rows to see what they represent
```

**Next Step:** Save df_full and create two new datasets with relevant data
- df_books - one row per book using the columns: ratings, title, author, year - used to analyze reviews, ratings, and engagement
- df_book_genres - one row per book AND genre combination using book ID and genre - used to compare engagement across genres

```python
from pathlib import Path

cleaned_path = Path.home() / "Desktop" / "Goodreads_Case_Study" / "cleaned_data"

cleaned_path.mkdir(parents=True, exist_ok=True)

df_full.to_parquet(
    cleaned_path / "goodreads_cleaned.parquet",
    index=False
) # save df_full as goodreads_cleaned.parquet
```

```python
print((cleaned_path / "goodreads_cleaned.parquet").exists()) # to verify it saved
```

```python
df_books = df_full[[
    "book_id",
    "work_id",
    "title",
    "first_author",
    "publication_year",
    "average_rating",
    "ratings_count",
    "reviews_count"
]].copy() # creating the table df_books as a new dataframe
```

```python
print(df_books.shape)
print(df_books["book_id"].nunique())
print(df_books["book_id"].duplicated().sum()) 
df_books.head() # checking new dataframe looks good and makes sense
```

```python
df_book_genres = (
    df_full[["book_id", "genres"]]
    .explode("genres")
    .rename(columns={"genres": "genre"})
    .dropna(subset=["genre"])
    .copy()
) # making df_book_genres using one genre per row with no books without classified genres
```

```python
df_book_genres.head(10) # making sure each row has a different single genre even if book_id is the same
```

```python
print(df_book_genres.shape)

print(df_book_genres["book_id"].nunique())

print(df_book_genres.duplicated(subset=["book_id", "genre"]).sum()) # looking at total book-genre combos, unique books, and duplicates
```

```python
df_books.to_parquet(
    cleaned_path / "books.parquet",
    index=False
)

df_book_genres.to_parquet(
    cleaned_path / "book_genres.parquet",
    index=False
) # saving new files
```

```python
print((cleaned_path / "books.parquet").exists())
print((cleaned_path / "book_genres.parquet").exists()) # double checking 
```
